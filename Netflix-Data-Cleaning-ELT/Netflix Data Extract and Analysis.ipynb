## 1. Load the Dataset

Load the Netflix titles CSV file into a Pandas DataFrame and inspect its dimensions and sample records.

```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python
import sqlalchemy as sal
import pyodbc

engine = sal.create_engine(
    "mssql+pyodbc://localhost/NetflixDB"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&Trusted_Connection=yes"
    "&TrustServerCertificate=yes"
)

conn = engine.connect()
print("Connected successfully!")
```

```python
df.to_sql('netflix_raw', con=conn, index=False, if_exists = 'append', dtype={'title': NVARCHAR(200)})
conn.close() #to handle the foreign characters added dtype;
```
